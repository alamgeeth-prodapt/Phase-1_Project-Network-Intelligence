```python
mapping = {
    "datetime":"datetime",
    "CellID":"grid_id",
    "countrycode":"country_code",
    "smsin":"sms_in",
    "smsout":"sms_out",
    "callin":"call_in",
    "callout":"call_out",
    "internet":"internet"
}

df_canonical = df.rename(columns=mapping)

df_canonical.dtypes
```

```python
df_canonical.head(1)
```

```python
import datetime
df_canonical["datetime"] = pd.to_datetime(df_canonical["datetime"])
print(df_canonical["datetime"].dt.hour.nunique())
```

```python
df_canonical["timestamp"] = df_canonical["datetime"].dt.time
df_canonical["date"] = df_canonical["datetime"].dt.date
df_canonical["hour"] = df_canonical["datetime"].dt.hour + 1
df_canonical["day"] = df_canonical["datetime"].dt.day

df_canonical.head(5)
```

```python
df_canonical = df_canonical.drop_duplicates()
df_canonical = df_canonical.fillna(0)

df_canonical.head(5)
```

```python
grid_count = (
    df_canonical.groupby(["hour","grid_id"])["country_code"].nunique()
)

print(grid_count)
```

```python
df_canonical["total_sms"] = df_canonical["sms_in"] + df_canonical["sms_out"]
```

```python
df_canonical.head(5)
```

```python
df_canonical["total_calls"] = df_canonical["call_in"] + df_canonical["call_out"]
```

```python
df_canonical["total_activity"] = df_canonical["total_calls"] + df_canonical["total_sms"] + df_canonical["internet"]
```
