```python
import pandas as pd
from sklearn.metrics import classification_report, accuracy_score, precision_score

# 1. Load your generated scores
scored = pd.read_csv("anomaly_scores.csv")

# 2. Define your ground-truth binary drop condition 
# (e.g., if actual traffic dropped significantly below rolling expected baseline)
# For example, let's define a true drop as an actual value dropping >50% below the rolling mean
scored['true_drop'] = (scored['internet_target'] < (scored['internet_roll_mean_24'] * 0.5)).astype(int)

# 3. Use your model's z-score anomaly flag as the predicted label
scored['predicted_drop'] = scored['is_anomaly'].astype(int)

# Drop rows where rolling baseline couldn't be calculated
eval_df = scored.dropna(subset=['true_drop', 'predicted_drop'])

# 4. Print the exact classification report
print(classification_report(eval_df['true_drop'], eval_df['predicted_drop'], digits=4))
```
